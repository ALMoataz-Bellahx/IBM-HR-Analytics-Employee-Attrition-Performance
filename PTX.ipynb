## --- 3. DATA CLEANING & ERROR HANDLING ---

```python
# Create a boolean mask of missing values, sum them up per column, and store the result in 'missing_values'
missing_values = df.isnull().sum()
```

```python
# Print the count of missing values to check if imputation or dropping rows is necessary (This dataset has 0 missing values)
print(missing_values)
```

```python
#Note on Error Handling: While there are no missing values, some columns have zero variance (only 1 unique value for all 1470 rows).
#Columns like 'EmployeeCount', 'Over18', and 'StandardHours' provide no analytical value and can cause plotting errors.
#'EmployeeNumber' is just a random ID tag and shouldn't be used for analysis.
```

```python
# Create a list containing the names of the columns that we want to remove from our dataset
columns_to_drop = ['EmployeeCount', 'Over18', 'StandardHours', 'EmployeeNumber']
```

```python
# Drop the specified columns from 'df' and save the cleaned data into a new DataFrame called 'df_cleaned'
df_cleaned = df.drop(columns=columns_to_drop)
```

```python
# Export the cleaned DataFrame to a new CSV file. 'index=False' prevents pandas from saving the row numbers as a new column.
df_cleaned.to_csv('Cleaned_HR_Employee_Attrition.csv', index=False)
```

## --- ADDITIONAL UNIVARIATE ANALYSIS ---

```python
# Q1: What is the distribution of Monthly Income among employees?
# Create a histogram using Plotly Express to visualize the spread of salaries
# 'x' is set to 'MonthlyIncome' to map the income brackets on the horizontal axis
# 'color_discrete_sequence' applies a specific hex color code to the bars for visual variety
fig_income_uni = px.histogram(df_cleaned, x='MonthlyIncome', title='Univariate: Distribution of Monthly Income', template='plotly_dark', color_discrete_sequence=['#457b9d'])
# Render and display the interactive income histogram
fig_income_uni.show()
```

```python
# Q2: How are employees distributed across Educational Fields?
# Create a bar chart showing the total count of employees originating from each education field
fig_edu = px.histogram(df_cleaned, x='EducationField', title='Univariate: Employees by Education Field', template='plotly_dark', color_discrete_sequence=['#1d3557'])
# Update the layout of the chart to sort the bars in descending order (tallest to shortest)
# This makes it instantly clear which educational background is most common
fig_edu.update_layout(xaxis={'categoryorder':'total descending'})
# Render and display the educational field chart
fig_edu.show()
```

```python
# Q3: What is the gender breakdown of the workforce?
# First, count the number of occurrences of each gender and use reset_index() to turn it into a DataFrame
gender_counts = df_cleaned['Gender'].value_counts().reset_index()
# Rename the columns of this new tiny DataFrame so Plotly can read them easily
gender_counts.columns = ['Gender', 'Count']
# Create a pie chart using the summarized counts to show the exact percentage breakdown
# 'color_discrete_sequence' assigns two specific hex colors for Male and Female slices
fig_gender = px.pie(gender_counts, values='Count', names='Gender', title='Univariate: Gender Breakdown', template='plotly_dark', color_discrete_sequence=['#2a9d8f', '#e76f51'])
# Render and display the gender pie chart
fig_gender.show()
```

```python
# Q4: What is the overall experience level of the workforce?
# Create a histogram focusing on 'TotalWorkingYears' to see if the workforce is junior or senior heavy
# 'nbins=30' explicitly tells Plotly to try and create 30 separate bars (bins) for granular detail
fig_working_years = px.histogram(df_cleaned, x='TotalWorkingYears', title='Univariate: Total Working Years Experience', template='plotly_dark', nbins=30)
# Render and display the working years histogram
fig_working_years.show()
```

```python
# Q5: What proportion of the workforce is required to travel for business?
# Create a bar chart mapping the frequency of business travel across the company
fig_travel = px.histogram(df_cleaned, x='BusinessTravel', title='Univariate: Business Travel Frequency', template='plotly_dark', color_discrete_sequence=['#a8dadc'])
# Sort the categories from most frequent to least frequent for clean readability
fig_travel.update_layout(xaxis={'categoryorder':'total descending'})
# Render and display the business travel chart
fig_travel.show()
```

```python
# Q6: How is the commute distance distributed among employees?
# Create a histogram to see if most employees live near the office or have long commutes
fig_distance = px.histogram(df_cleaned, x='DistanceFromHome', title='Univariate: Commute Distance (From Home)', template='plotly_dark', color_discrete_sequence=['#8d99ae'])
# Render and display the commute distance histogram
fig_distance.show()
```

```python
# Q7: What is the breakdown of Marital Status in the company?
# Count the occurrences of 'Single', 'Married', and 'Divorced', resetting the index for Plotly
marital_counts = df_cleaned['MaritalStatus'].value_counts().reset_index()
# Rename the columns of the output for clarity
marital_counts.columns = ['MaritalStatus', 'Count']
# Create a donut chart (a pie chart with a hole in the middle) for a modern aesthetic
# The 'hole=0.4' parameter cuts out the middle 40% of the pie chart
fig_marital = px.pie(marital_counts, values='Count', names='MaritalStatus', title='Univariate: Marital Status Distribution', template='plotly_dark', hole=0.4)
# Render and display the marital status donut chart
fig_marital.show()
```

```python
# Q8: How long do employees typically stay at this company?
# Create a histogram focused strictly on 'YearsAtCompany' to visualize overall employee tenure
fig_tenure = px.histogram(df_cleaned, x='YearsAtCompany', title='Univariate: Years at the Company (Tenure)', template='plotly_dark', color_discrete_sequence=['#d62828'])
# Render and display the company tenure histogram
fig_tenure.show()
```
